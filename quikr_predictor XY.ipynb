## Cleaning

```python
backup=car.copy()
```

```python
car=car[car["year"].str.isnumeric()].copy()
car["year"]=car["year"].astype(int)
```

```python
car.info()
```

```python
car['kms_driven'] = car['kms_driven'].astype('string')

car['kms_driven'] = car['kms_driven'].str.replace(',', '', regex=False)
car['kms_driven'] = car['kms_driven'].str.replace(
    r'\s*kms\s*', '', regex=True
)

car['kms_driven'] = pd.to_numeric(
    car['kms_driven'], errors='coerce'
)

print(car['kms_driven'].dtype)
print(car['kms_driven'].isna().sum())
```

```python
car=car[car["Price"]!="Ask For Price"]
```

```python
car["Price"] = (
    car["Price"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.replace(" ", "", regex=False)
)
```

```python
car["Price"] = pd.to_numeric(car["Price"], errors="coerce")
```

```python
car.info()
```

```python
car=car[~car['fuel_type'].isna()]
```

```python
car['fuel_type'].unique()
```

```python
car=car.dropna()
```

```python
car=car.drop_duplicates()
```

```python
car=car[car['Price']<6e6]
car=car[car['kms_driven']<=250000]
car=car[~((car['kms_driven']<1000)&(car['year']<=2016))]
```

```python
car['name']=car['name'].str.split(' ').str.slice(0,3).str.join(' ')
car['kms_driven']=car['kms_driven'].astype(int)
```

```python
car=car.reset_index(drop=True)
```

```python
car.info()
```

```python
car.describe()
```

```python
print(car['fuel_type'].value_counts(dropna=False))
```

```python
car.to_csv('cleaned_quikr_car.csv', index=False)
```

## Model

```python
x=car.drop(columns=['Price'])
y= car['Price']
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
x, y, test_size=0.2, random_state=42
)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.preprocessing import OneHotEncoder, FunctionTransformer
from sklearn.compose import make_column_transformer, TransformedTargetRegressor
from sklearn.pipeline import make_pipeline
```

```python
column_trans = make_column_transformer(
    (OneHotEncoder(handle_unknown='ignore'),
     ['name', 'company', 'fuel_type']),
    (FunctionTransformer(np.log1p), ['kms_driven']),
    remainder='passthrough'
)
```

```python
lr=LinearRegression()
```

```python
pipe=make_pipeline(column_trans,lr)
pipe=TransformedTargetRegressor(regressor=pipe,func=np.log,inverse_func=np.exp)
```

```python
pipe.fit(x_train, y_train)

y_pred = pipe.predict(x_test)
```

```python
r2_score(y_test,y_pred)
```

```python
from sklearn.model_selection import cross_val_score, RepeatedKFold

cv = RepeatedKFold(n_splits=5, n_repeats=5, random_state=42)
cross_val_score(pipe, x, y, cv=cv, scoring='r2').mean()
```
